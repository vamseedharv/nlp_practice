Wordnet Lemmatizer
Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −

⚡ Key Difference from Stemming
- Stemming: crude chopping of word endings (studies → studi).
- Lemmatization: linguistically informed, produces valid dictionary words (studies → study).

🌱 Types of Lemmatization
- Dictionary-based (Lexicon-driven)
    - Uses a predefined dictionary or lexicon to map inflected forms to their lemmas.
    - Example: running → run, better → good (because the dictionary knows the irregular mapping).
    - Tools: WordNetLemmatizer in NLTK, spaCy’s lemmatizer.
- Rule-based
    - Applies morphological rules to strip suffixes/prefixes and reduce words to their lemma.
    - Example: studies → study (rule: remove -ies → y).
    - Often language-specific, since rules differ across languages.
- Hybrid (Dictionary + Rules)
    - Combines dictionary lookups with rules for cases not covered in the lexicon.
    - Example: spaCy lemmatizer uses both approaches.
- Statistical / Machine Learning-based
    - Uses trained models to predict the lemma based on context.
    - Example: Neural lemmatizers in modern NLP frameworks (like Stanza or UDPipe).
    - Handles ambiguous cases better (e.g., saw → see vs. saw → saw depending on context).
- Contextual Lemmatization
    - Goes beyond morphology and considers sentence meaning.
    - Example: flies → fly (noun) vs. flies → fly (verb).
    - Requires POS tagging or dependency parsing before lemmatization.

👉 In practice:
- WordNetLemmatizer (NLTK) → dictionary-based, requires POS tags for accuracy.
- spaCy Lemmatizer → hybrid, fast and practical.
- Stanza / UDPipe → statistical/contextual, better for multilingual NLP.



```python
# Lemmatization in NLTK
# Lemmatization is the process of reducing a word to its base or root form, known as a lemma.
# Unlike stemming, which simply cuts off prefixes or suffixes, lemmatization considers the context and the morphological analysis of the word to ensure that the root form is a valid word in the language.
# In this example, we will use the WordNetLemmatizer from the NLTK library to perform lemmatization on a list of words.
from nltk.stem import WordNetLemmatizer
import nltk
# nltk.download('wordnet') # Uncomment this line to download the necessary WordNet data
lemmatizer = WordNetLemmatizer()
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
for word in words:
    print(f"{word} -> {lemmatizer.lemmatize(word, pos='v')}")  # pos='v' indicates that we are lemmatizing verbs
```
