**Tokenization in NLTK** is the process of breaking text into smaller units called **tokens** (words, sentences, or subwords). It’s the **first step** in most NLP pipelines.

---

# 🧠 Types of Tokenization in NLTK

## 1️⃣ Sentence Tokenization

Splits text into sentences.

```python
import nltk
from nltk.tokenize import sent_tokenize

nltk.download('punkt')

text = "Hello Vamsee. You are learning NLP. That's great!"

sentences = sent_tokenize(text)
print(sentences)
```

### Output:

```python
['Hello Vamsee.', 'You are learning NLP.', "That's great!"]
```

---

## 2️⃣ Word Tokenization

Splits sentences into words/tokens.

```python
from nltk.tokenize import word_tokenize

text = "Vamsee is learning NLP using Python."
words = word_tokenize(text)

print(words)
```

### Output:

```python
['Vamsee', 'is', 'learning', 'NLP', 'using', 'Python', '.']
```

---

## 3️⃣ Punctuation-based Tokenization

Simple split based on spaces & punctuation.

```python
from nltk.tokenize import wordpunct_tokenize

text = "Hello! How's NLP going?"
tokens = wordpunct_tokenize(text)

print(tokens)
```

### Output:

```python
['Hello', '!', 'How', "'", 's', 'NLP', 'going', '?']
```

---

## 4️⃣ Custom Tokenization (Regex)

You can define your own rules.

```python
from nltk.tokenize import RegexpTokenizer

tokenizer = RegexpTokenizer(r'\w+')
text = "Vamsee's NLP journey started in 2025!"

tokens = tokenizer.tokenize(text)
print(tokens)
```

### Output:

```python
['Vamsee', 's', 'NLP', 'journey', 'started', 'in', '2025']
```

# ⚙️ Why Tokenization Matters

Tokenization helps machines understand text structure:

* Converts raw text → structured input
* Required before:

  * POS tagging
  * Named Entity Recognition
  * Sentiment analysis
  * LLM pipelines

# 🔍 Comparison of Tokenizers

| Tokenizer            | Best For           | Behavior             |
| -------------------- | ------------------ | -------------------- |
| `sent_tokenize`      | Sentence splitting | Language-aware       |
| `word_tokenize`      | General NLP        | Smart, context-aware |
| `wordpunct_tokenize` | Simple splitting   | Breaks punctuation   |
| `RegexpTokenizer`    | Custom rules       | Fully flexible       |

````python
text = """Don't hesitate—Mr. Smith's well-known email (example@test.com) arrived at 3:45 p.m.""" 
````

````python
# Tokenization example using NLTK
# Tokenization is the process of breaking down text into smaller units, such as words or sentences. 
# In this example, we will use the NLTK library to tokenize a simple sentence into words.
# We will use the word_tokenize function from NLTK to perform the tokenization.
from nltk.tokenize import word_tokenize
import nltk
# nltk.download('punkt') # Uncomment this line to download the necessary tokenizer models
tokens = word_tokenize(text)
for token in tokens:
    print(token)
````

````python
# Another tokenization example using NLTK's wordpunct_tokenize
# The wordpunct_tokenize function splits the text into tokens based on both whitespace and punctuation.
# This can be useful for certain applications where you want to keep punctuation as separate tokens.
# Note that this will treat punctuation as separate tokens, which may or may not be desirable depending on your use case.
# For example, "Hello, world!" would be tokenized into ["Hello", ",", "world", "!"].

from nltk.tokenize import wordpunct_tokenize
import nltk
# nltk.download('punkt') # Uncomment this line to download the necessary tokenizer models
tokens = wordpunct_tokenize(text)
for token in tokens:
    print(token)
    
````

````python
# Another tokenization example using NLTK's TreebankWordTokenizer
# The TreebankWordTokenizer is a more sophisticated tokenizer that is designed to mimic the tokenization used in the Penn Treebank.
# It handles contractions and punctuation in a way that is more consistent with how humans would tokenize text.
# For example, "Don't" would be tokenized into ["Do", "n't"] and "Hello, world!" would be tokenized into ["Hello", ",", "world", "!"].
# This tokenizer is often used in NLP applications where accurate tokenization is important, such as in parsing or part-of-speech tagging.
# Note that you may need to download the necessary tokenizer models using nltk.download('punkt') before using this tokenizer.
# The TreebankWordTokenizer is particularly useful for tokenizing English text, as it is designed to handle the specific quirks of English tokenization, such as contractions and punctuation.
# In this example, we will use the TreebankWordTokenizer to tokenize the same text as before and see how it handles the tokenization differently from the word_tokenize and wordpunct_tokenize functions.
# The TreebankWordTokenizer will split contractions into separate tokens and will also handle punctuation in a way that is more consistent with how humans would tokenize text.
# For example, "Don't" would be tokenized into ["Do", "n't"] and "Hello, world!" would be tokenized into ["Hello", ",", "world", "!"].
from nltk.tokenize import TreebankWordTokenizer
import nltk
# nltk.download('punkt') # Uncomment this line to download the necessary tokenizer models
treebankWordTokenizer = TreebankWordTokenizer()
tokens = treebankWordTokenizer.tokenize(text)
for token in tokens:
    print(token)
````

- RegexpTokenizer
-   Lets you define custom regular expressions for splitting text.
-   Example: split only on whitespace or keep punctuation attached.
- BlanklineTokenizer
-   Splits text into paragraphs based on blank lines.
-   Useful for document-level processing.
- WhitespaceTokenizer
-   Splits strictly on spaces and tabs.
-   Doesn’t handle punctuation or contractions intelligently.
- Stanford Tokenizer (via NLTK wrapper)
-   Uses Stanford NLP’s Java-based tokenizer.
-   More linguistically sophisticated, but requires external setup


````python
# Another tokenization example using NLTK's sent_tokenize
# The sent_tokenize function is used to tokenize text into sentences.
# This can be useful for applications where you want to process text at the sentence level, such as in sentiment analysis or summarization.
# The sent_tokenize function uses a pre-trained model to identify sentence boundaries, which can be more accurate than simply splitting on punctuation.
from nltk.tokenize import sent_tokenize
import nltk
# nltk.download('punkt') # Uncomment this line to download the necessary tokenizer models
sentences = sent_tokenize(text)
for sentence in sentences:
    print(sentence)
````

````python
# Another tokenization example using NLTK's RegexpTokenizer
# The RegexpTokenizer allows you to specify a regular expression pattern to define how the text should be tokenized.
# This can be useful for custom tokenization needs, such as extracting specific types of tokens or ignoring certain characters.
# In this example, we will use a regular expression pattern to tokenize the text into words while ignoring punctuation.
# The regular expression pattern \w+ matches sequences of word characters (letters, digits, and underscores), effectively ignoring punctuation.
from nltk.tokenize import RegexpTokenizer
import nltk
# nltk.download('punkt') # Uncomment this line to download the necessary tokenizer models
#- It finds whole words made up of letters, digits, or underscores.
#- It ignores punctuation, spaces, and symbols.
regexpTokenizer = RegexpTokenizer(r'\w+')
tokens = regexpTokenizer.tokenize(text)
for token in tokens:
    print(token)
````

````python
# Another tokenization example using NLTK's BlanklineTokenizer
# The BlanklineTokenizer is used to tokenize text based on blank lines.
# This can be useful for applications where you want to process text at the paragraph level, such as in document summarization or topic modeling.
# The BlanklineTokenizer will split the text into tokens based on blank lines, treating each block of text separated by blank lines as a separate token.
from nltk.tokenize import BlanklineTokenizer
import nltk
# nltk.download('punkt') # Uncomment this line to download the necessary tokenizer models
blanklineTokenizer = BlanklineTokenizer()
text = """Hello Welcome,to Vamseedhar's NLP practice sessions.








Please keep practicing! to become expert in NLP."""
tokens = blanklineTokenizer.tokenize(text)
for token in tokens:
    print(token)

#sent_tokenize and BlanklineTokenizer are used to tokenize text into sentences and paragraphs, respectively.
# The sent_tokenize function uses a pre-trained model to identify sentence boundaries, while the BlanklineTokenizer splits text based on blank lines.
# The choice between these tokenizers depends on the level of granularity you want in your tokenization. If you want to process text at the sentence level, sent_tokenize is a good choice. If you want to process text at the paragraph level, BlanklineTokenizer is more appropriate.
# In practice, you may want to use a combination of different tokenizers depending on the specific requirements of your NLP application. For example, you might use sent_tokenize to split text into sentences and then use word_tokenize to further tokenize each sentence into words.
# It's important to choose the right tokenizer for your specific use case, as different tokenizers may produce different results based on how they handle punctuation, contractions, and other aspects of text.
# In summary, NLTK provides a variety of tokenization functions that can be used to break down text into smaller units, such as words, sentences, or paragraphs. The choice of tokenizer depends on the specific requirements of your NLP application and the level of granularity you want in your tokenization.
# --- IGNORE ---
# In this example, we have seen how to use different tokenization functions from the NLTK library to tokenize text into words, sentences, and paragraphs. Each tokenizer has its own strengths and weaknesses, and the choice of tokenizer depends on the specific requirements of your NLP application. By understanding the different tokenization options available in NLTK, you can choose the most appropriate tokenizer for your needs and improve the accuracy and effectiveness of your NLP tasks.
# --- IGNORE ---
# In conclusion, tokenization is a fundamental step in natural language processing that involves breaking down text into smaller units. NLTK provides a variety of tokenization functions that can be used to tokenize text into words, sentences, and paragraphs. By choosing the right tokenizer for your specific use case, you can improve the accuracy and effectiveness of your NLP applications. Whether you are working on sentiment analysis, summarization, or any other NLP task, understanding how to use tokenization effectively is essential for success.
# --- IGNORE ---
````
