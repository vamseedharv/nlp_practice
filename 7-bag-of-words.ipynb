**Bag of Words (BoW)** is a simple technique in Natural Language Processing used to convert text into numerical data so machines can understand it.

---

### 🔹 Core idea

It represents a piece of text by:

* Ignoring grammar and word order
* Keeping only **word frequency (counts)**

Think of it as putting all words into a “bag” 👜 and just counting them.

---

### 🔹 Example

Suppose you have two sentences:

1. "I love AI"
2. "I love coding"

**Step 1: Build vocabulary**

```
[I, love, AI, coding]
```

**Step 2: Convert sentences into vectors**

| Sentence      | I | love | AI | coding |
| ------------- | - | ---- | -- | ------ |
| I love AI     | 1 | 1    | 1  | 0      |
| I love coding | 1 | 1    | 0  | 1      |

---

### 🔹 Why use it?

* Converts text into numbers for ML models
* Simple and easy to implement
* Works well for basic text tasks like spam detection

---

### 🔹 Limitations

* ❌ Ignores word order (“not good” vs “good not” look same)
* ❌ No understanding of meaning/context
* ❌ Large vocabulary → high dimensional vectors

---

### 🔹 Variations

* **Binary BoW** → just 0/1 (word present or not)
* **Term Frequency (TF)** → counts of words
* Often extended into **TF-IDF (Term Frequency–Inverse Document Frequency)**

---

### 🔹 Example in Python

Using scikit-learn:

```python
from sklearn.feature_extraction.text import CountVectorizer

corpus = ["I love AI", "I love coding"]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

---

### 🔹 When to use

* Text classification (spam detection, sentiment analysis)
* Document similarity
* Basic NLP tasks where simplicity matters

````python
## Data Cleaning And Preprocessing
import re
import nltk
# nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
ps=PorterStemmer()
sentence="I loved the movie! It was fantastic and thrilling."
review=re.sub('[^a-zA-z]',' ',sentence)
review=review.lower()
review=review.split()
review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
review=' '.join(review)
print(review)
# ngram_range=(1,2) means that we want to create both unigrams (single words) and bigrams (pairs of consecutive words).
cv=CountVectorizer(max_features=100,binary=True) # max_features=100 limits the number of features to 100, and binary=True indicates that we want to create a binary matrix (1 for presence, 0 for absence of a word).
X=cv.fit_transform([review]).toarray()
print(X)
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, 
    formatter=dict(float=lambda X: "%.3g" % X))
print(cv.get_feature_names_out())
print(cv.vocabulary_)
````
