In **NLTK (Natural Language Toolkit)**, *parts of speech (POS)* tagging is the process of labeling words in a sentence with their grammatical roles like noun, verb, adjective, etc.

---

# 🧠 Common Parts of Speech in NLTK

NLTK uses the **Penn Treebank POS tags**. Here are the most important ones:

| Tag     | Meaning                | Example     |
| ------- | ---------------------- | ----------- |
| **NN**  | Noun (singular)        | dog, car    |
| **NNS** | Noun (plural)          | dogs        |
| **NNP** | Proper noun            | India, John |
| **VB**  | Verb (base)            | run         |
| **VBD** | Verb (past)            | ran         |
| **VBG** | Verb (gerund)          | running     |
| **VBN** | Verb (past participle) | eaten       |
| **VBP** | Verb (present)         | run         |
| **VBZ** | Verb (3rd person)      | runs        |
| **JJ**  | Adjective              | big         |
| **JJR** | Comparative adjective  | bigger      |
| **JJS** | Superlative adjective  | biggest     |
| **RB**  | Adverb                 | quickly     |
| **IN**  | Preposition            | in, on      |
| **DT**  | Determiner             | the, a      |
| **PRP** | Pronoun                | he, she     |
| **CC**  | Conjunction            | and, but    |

---

# ⚙️ How POS Tagging Works in NLTK

### Step-by-step example:

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import pos_tag

# Download required resources (run once)
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

text = "Vamsee is learning NLP using Python"

# Tokenize sentence
tokens = word_tokenize(text)

# POS tagging
tagged = pos_tag(tokens)

print(tagged)
```

### Output:

```python
[('Vamsee', 'NNP'),
 ('is', 'VBZ'),
 ('learning', 'VBG'),
 ('NLP', 'NNP'),
 ('using', 'VBG'),
 ('Python', 'NNP')]
```

---

# 🔍 What’s Happening Here?

* `word_tokenize()` → breaks sentence into words
* `pos_tag()` → assigns grammatical labels
* Uses a **pre-trained statistical model** (not rules only)

**Stop words in NLTK** are commonly used words (like *the, is, in, and*) that are usually **removed during text processing** because they don’t add much meaning.

---

# 🧠 What are Stop Words?

Examples of English stop words:

* the, is, am, are
* in, on, at
* and, but, or
* he, she, it

👉 These words appear frequently but carry **little semantic value** in many NLP tasks.

---

# ⚙️ How to Use Stop Words in NLTK

## Step 1: Import and Download

```python id="s8d9sj"
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
```

---

## Step 2: View Stop Words

```python id="k3jd82"
stop_words = set(stopwords.words('english'))
print(list(stop_words)[:10])
```

---

## Step 3: Remove Stop Words from Text

```python id="x8s7dj"
from nltk.tokenize import word_tokenize

nltk.download('punkt')

text = "Vamsee is learning NLP using Python and building cool projects"

words = word_tokenize(text)

filtered_words = [word for word in words if word.lower() not in stop_words]

print(filtered_words)
```

### Output:

```python id="p92k3j"
['Vamsee', 'learning', 'NLP', 'using', 'Python', 'building', 'cool', 'projects']
```

---

# 🔍 What’s Happening?

* Convert words → lowercase (`word.lower()`)
* Compare with stop word list
* Remove matches
* Keep meaningful words only

---

# 🚀 Why Remove Stop Words?

Useful for:

* Search engines
* Text classification
* Sentiment analysis
* Topic modeling

👉 Helps reduce noise and improves model performance.

---

# ⚠️ When NOT to Remove Stop Words

Sometimes stop words are important:

* **Sentiment analysis**

  * "not good" → removing "not" changes meaning ❌
* **Question answering**
* **Language generation (LLMs)**

---

# 🔄 Alternatives in Modern NLP

* spaCy → built-in stop word handling
* Hugging Face Transformers → often **don’t remove stop words** (models learn context)

---

# 🧩 Quick Summary

* Stop words = frequent, low-meaning words
* Use `stopwords.words('english')`
* Remove using filtering
* Be careful: sometimes they matter!


````python
## Speech Of DR APJ Abdul Kalam
paragraph = """I have three visions for India. In 3000 years of our history, people from all over
               the world have come and invaded us, captured our lands, conquered our minds.
               From Alexander onwards, the Greeks, the Turks, the Moguls, the Portuguese, the British,
               the French, the Dutch, all of them came and looted us, took over what was ours.
               Yet we have not done this to any other nation. We have not conquered anyone.
               We have not grabbed their land, their culture,
               their history and tried to enforce our way of life on them.
               Why? Because we respect the freedom of others.That is why my
               first vision is that of freedom. I believe that India got its first vision of
               this in 1857, when we started the War of Independence. It is this freedom that
               we must protect and nurture and build on. If we are not free, no one will respect us.
               My second vision for India’s development. For fifty years we have been a developing nation.
               It is time we see ourselves as a developed nation. We are among the top 5 nations of the world
               in terms of GDP. We have a 10 percent growth rate in most areas. Our poverty levels are falling.
               Our achievements are being globally recognised today. Yet we lack the self-confidence to
               see ourselves as a developed nation, self-reliant and self-assured. Isn’t this incorrect?
               I have a third vision. India must stand up to the world. Because I believe that unless India
               stands up to the world, no one will respect us. Only strength respects strength. We must be
               strong not only as a military power but also as an economic power. Both must go hand-in-hand.
               My good fortune was to have worked with three great minds. Dr. Vikram Sarabhai of the Dept. of
               space, Professor Satish Dhawan, who succeeded him and Dr. Brahm Prakash, father of nuclear material.
               I was lucky to have worked with all three of them closely and consider this the great opportunity of my life.
               I see four milestones in my career""" 
````

````python
# applying stemming and stop words removal using words tokenization and sentences tokenization
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
import nltk
#nltk.download('punkt')
#nltk.download('stopwords')
stemmer = PorterStemmer()
# sentences tokenization
sentences = nltk.sent_tokenize(paragraph)
for i in range(len(sentences)):
    # print(sentences[i]) 
    # words tokenization   
    words = nltk.word_tokenize(sentences[i])
    # stemming and stop words removal from each sentence
    words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    # joining the words to form the sentence again
    sentences[i] = ' '.join(words)  
    # printing the sentences after stemming and stop words removal
    print(sentences[i])

# print(sentences)
````

````python
# applying stop words removal using words tokenization and sentences tokenization and snowball stemming
from nltk.stem import SnowballStemmer   
snowball_stemmer = SnowballStemmer('english')
# sentences tokenization
sentences = nltk.sent_tokenize(paragraph)
for i in range(len(sentences)):
    # print(sentences[i]) 
    # words tokenization   
    words = nltk.word_tokenize(sentences[i])
    # stemming and stop words removal from each sentence
    words = [snowball_stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    # joining the words to form the sentence again
    sentences[i] = ' '.join(words)  
    # printing the sentences after stemming and stop words removal
    print(sentences[i])
````

````python
# appling stop words removal using words tokenization and sentences tokenization and lemmatization
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
# sentences tokenization
sentences = nltk.sent_tokenize(paragraph)
for i in range(len(sentences)):
    # print(sentences[i]) 
    # words tokenization   
    words = nltk.word_tokenize(sentences[i])
    # lemmatization and stop words removal from each sentence
    words = [lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words('english'))]
    # joining the words to form the sentence again
    sentences[i] = ' '.join(words)  
    # printing the sentences after lemmatization and stop words removal
    print(sentences[i])
````
