## 🔵 Continuous Bag of Words (CBOW) — Detailed Explanation

**Continuous Bag of Words (CBOW)** is a neural network model used in Natural Language Processing to learn **word embeddings** (dense vector representations of words). It is one of the two main architectures of Word2Vec, introduced by Tomas Mikolov and his team at Google.

---

# 🔹 1. Core Idea

CBOW predicts a **target word** using its **surrounding context words**.

👉 In simple terms:

> “Given nearby words, guess the missing word.”

---

# 🔹 2. Example

Sentence:
**“I love artificial intelligence”**

If window size = 2:

* Context words: **[I, love, intelligence]**
* Target word: **artificial**

So the model learns:

```
[I, love, intelligence] → artificial
```

---

# 🔹 3. Architecture (How CBOW works)

CBOW is a **shallow neural network** with 3 layers:

### 🔸 1. Input Layer

* Takes context words
* Each word is one-hot encoded

---

### 🔸 2. Hidden Layer

* All input vectors are **averaged (or summed)**
* Multiplied by weight matrix
* Produces a **dense vector (embedding)**

---

### 🔸 3. Output Layer

* Uses **softmax function** to predict probability of each word
* Chooses the most likely target word

---

### 📊 Flow diagram (conceptual)

```
Context words → One-hot → Average → Hidden layer → Softmax → Target word
```

---

# 🔹 4. Mathematical intuition (simplified)

Let:

* Vocabulary size = V
* Embedding size = N

### Steps:

1. Convert context words into one-hot vectors
2. Compute average:

   ```
   h = (x1 + x2 + ... + xn) / n
   ```
3. Multiply with weights:

   ```
   h = Wᵀ x
   ```
4. Apply softmax:

   ```
   P(word) = exp(score) / sum(exp(all scores))
   ```

---

# 🔹 5. Training process

CBOW learns by adjusting weights to reduce prediction error:

* Uses **backpropagation**
* Optimized with:

  * Negative sampling OR
  * Hierarchical softmax

---

# 🔹 6. Key features

✔ Fast to train
✔ Works well for frequent words
✔ Produces meaningful embeddings
✔ Captures semantic similarity

---

# 🔹 7. Advantages

* Efficient for large datasets
* Smooths over context (averaging reduces noise)
* Good general-purpose embeddings

---

# 🔹 8. Disadvantages

* ❌ Ignores word order (like BoW)
* ❌ Less effective for rare words
* ❌ Context averaging loses some meaning

---

# 🔹 9. CBOW vs Skip-gram (quick insight)

| Feature    | CBOW                | Skip-gram           |
| ---------- | ------------------- | ------------------- |
| Predicts   | Target from context | Context from target |
| Speed      | Faster              | Slower              |
| Rare words | Worse               | Better              |

---

# 🔹 10. Real-world applications

* Word similarity (king ≈ queen 👑)
* Text classification
* Recommendation systems
* Pretraining embeddings for deep learning models

---

# 🔹 11. Example in Python (using Word2Vec)

Using gensim:

```python id="q3z8lm"
from gensim.models import Word2Vec

sentences = [["I", "love", "AI"], ["AI", "is", "powerful"]]

model = Word2Vec(sentences, vector_size=50, window=2, sg=0)  # sg=0 → CBOW

print(model.wv['AI'])
```

---

# 🔹 12. Key takeaway

👉 **CBOW learns meaning by predicting a word from its context**
👉 It converts words into **dense vectors** that capture relationships
