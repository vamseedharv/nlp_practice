Named Entity Recognition (NER) is a Natural Language Processing (NLP) technique used to identify and classify key information (entities) in text into predefined categories.
🔍 What NER Does
NER scans text and extracts entities such as:

Persons (e.g., Satya Nadella)
Organizations (e.g., Microsoft)
Locations (e.g., Bangalore)
Dates & Times (e.g., 17 March 2026)
Numerical values (e.g., ₹5,000, 10%)
Products, Events, etc. (advanced NER systems)


📘 Example
Input:
“Satya Nadella visited Bangalore in March 2026 for a Microsoft event.”
Output (NER):

Satya Nadella → PERSON
Bangalore → LOCATION
March 2026 → DATE
Microsoft → ORGANIZATION


🛠️ Where NER is used

Search engines
Chatbots
Resume parsing
Customer support automation
Information extraction
Document summarization

```python
sentence="Barack Obama was the 44th President of the United States."
```

```python
# Named Entity Recognition (NER)
# https://www.nltk.org/book/ch07.html
# NER is the process of identifying and classifying named entities in text into predefined categories such as person names, organizations, locations, dates, etc. NLTK provides a built-in function for performing NER using a pre-trained model.
# Here's how you can use NLTK to perform Named Entity Recognition:
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')

# words=nltk.word_tokenize(sentence)
# tag_elements=nltk.pos_tag(words)
# Sample text
text = "Barack Obama was the 44th President of the United States."
# Tokenize the text
tokens = word_tokenize(text)
# Perform part-of-speech tagging
pos_tags = pos_tag(tokens)
# Perform Named Entity Recognition
named_entities = ne_chunk(pos_tags)
# Print the named entities
print(named_entities)
# In this example, we first tokenize the input text into words, then we perform part-of-speech tagging to identify the grammatical role of each word. Finally, we use the `ne_chunk` function to identify named entities in the text. The output will show the named entities along with their corresponding categories (e.g., PERSON, ORGANIZATION, GPE for geopolitical entity).
# Note: The `ne_chunk` function uses a pre-trained model, so it may not always be accurate, especially for less common entities or in contexts that differ from the training data. For more advanced NER tasks, you might want to consider using libraries like SpaCy or Hugging Face's Transformers, which offer more powerful and customizable NER models.
```
